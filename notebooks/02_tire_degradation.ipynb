# 02 Tire degradation

## Goal

Fit and check constructor-level tire degradation models per compound and circuit.

## Inputs

Clean laps and the tire features from src/features/tire_features.py.

## Outputs

Degradation slopes per constructor and compound, plus MAE and slope error on 2024.

## Step checklist

- [ ] Build the tire features per stint (tyre age, fuel-corrected lap time, stint delta).
- [ ] Fit the linear slope baseline per stint.
- [ ] Fit the main model with circuit, compound and constructor features.
- [ ] Compare the main model with the naive baseline on 2024 validation.
- [ ] Save the model artifact under models/.
- [ ] Note the slopes for the strategy notebook.
