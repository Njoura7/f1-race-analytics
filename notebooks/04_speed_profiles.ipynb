# 04 Speed profiles

## Goal

Compare constructor speed profiles at 3 chosen circuits for the 2024 fastest laps.

## Inputs

Car telemetry and circuit info from FastF1 for the 3 circuits, resampled by src/features/telemetry_features.py.

## Outputs

Speed traces, corner minimum speeds, top speeds and cumulative delta time between constructors.

## Step checklist

- [ ] Resample telemetry on a common distance grid.
- [ ] Segment the corners from the circuit info.
- [ ] Compute minimum corner speeds and top speeds per constructor.
- [ ] Build the cumulative delta time between constructors.
- [ ] Compare the constructors at the 3 circuits.
