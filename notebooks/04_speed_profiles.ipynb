# 04 Speed profiles

## Goal

Compare constructor speed profiles at the same circuit.

## Inputs

Car telemetry and circuit info from FastF1, resampled by src/features/telemetry_features.py.

## Outputs

Speed traces, corner speeds, top speeds and cumulative delta time between constructors.

## Step checklist

- [ ] Resample telemetry on a common distance grid.
- [ ] Segment corners from the circuit info.
- [ ] Compute minimum corner speeds and top speeds per constructor.
- [ ] Build the cumulative delta time between constructors.
- [ ] Compare constructors per circuit and note circuit similarities.
