# 03 Pit strategy

## Goal

Simulate one-stop and two-stop pit strategies and compare with what teams did in 2025.

## Inputs

Degradation model from notebook 02 and the strategy features from src/features/strategy_features.py.

## Outputs

Recommended strategies, time deltas and stop-lap agreement against 2025.

- [ ] Compare it with the actual 2025 strategies at the 3 circuits.
- [ ] Report the time delta and stop-lap agreement.
