# 03 Pit strategy

## Goal

Simulate pit strategies and compare the recommended one with what teams did in 2025.

## Inputs

Degradation model from notebook 02 and the strategy features from src/features/strategy_features.py.

## Outputs

Recommended strategies, time deltas and stop-lap agreement against 2025.

## Step checklist

- [ ] Compute pit loss per circuit and safety car rates from 2023 and 2024.
- [ ] Simulate candidate strategies with Monte Carlo on 2024 validation.
- [ ] Pick the strategy with the smallest total race time.
- [ ] Compare it with the actual 2025 strategies.
- [ ] Report the time delta and stop-lap agreement.
